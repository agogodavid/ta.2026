# Introduction to Data Models in R

In this notebook, we will explore the concept of data models, starting from the basic idea of a table, moving on to joins, and finally discussing semi-structured formats like JSON. All examples will be demonstrated using R.

## 1. What is a Data Table?

A data table is a structured way to organize data in rows and columns, similar to a spreadsheet. Each row represents an observation, and each column represents a variable or attribute.

```r
# Let's create a simple data frame (table) in R
customers <- data.frame(
  customer_id = 1:4,
  name = c("Alice", "Bob", "Charlie", "Diana"),
  age = c(20, 21, 19, 22)
)

# Display the data frame
customers
```

## 2. Relational Data and Joins

Often, data is split across multiple tables to reduce redundancy. These tables can be linked using keys (such as an ID column). Combining data from multiple tables is called a **join**.

```r
# Create another table: student grades
rentals <- data.frame(
  customer_id = c(1, 2, 2, 3, 4, 4),
  course = c("Odyssey", "The Green Window", "Vickies Truth", "Odyssey", "Working Windows", "Time Before Time"),
  payment = c(1.90, 2.85, 2.08, 1.92, 3.95, 4.89)
)

rentals
```

### Joining Tables in R

We can use the `dplyr` package to join tables. Let's join the `students` and `grades` tables to see each student's grades along with their names and ages.

```r
library(dplyr)

# Perform an inner join on the 'id' column
customer_rentals <- inner_join(customers, rentals, by = "customer_id")

customer_rentals
```

## 3. Semi-Structured Data: JSON

Not all data fits neatly into tables. **Semi-structured data** like JSON allows for more flexible data representation, where each record can have a different structure.

```r
# Example of JSON data in R
library(jsonlite)

# Create a list with varying structure
customer_profile <- list(
  list(id = 1, name = "Alice", favorite_genre = c("Horror", "Drama")),
  list(id = 2, name = "Bob", collector = TRUE),
  list(id = 3, name = "Charlie", age = 19, address = list(city = "Springfield", zip = "12345"))
)

# Convert to JSON
json_data <- toJSON(customer_profile, pretty = TRUE)
cat(json_data)
```

### Reading and Manipulating JSON in R

R can read and manipulate JSON data using the `jsonlite` package. Let's parse the JSON data back into an R object.

```r
# Parse JSON back to R list
parsed_profiles <- fromJSON(json_data, simplifyDataFrame = FALSE)

# Display the parsed data
parsed_profiles
```

JSON allows us to store nested and variable-length data, such as a list of courses for each student. This is useful when the structure of the data can change or when relationships are more complex.

## 5. Summary

- **Tables** organize data into rows and columns.
- **Relationships** connect data across tables using keys.
- **Joins** allow us to combine data from multiple tables.
- **Semi-structured formats** like JSON provide flexibility for complex or variable data.

Understanding these concepts is essential for working with data in both traditional databases and modern data systems.
